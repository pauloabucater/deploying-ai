```python
import requests
import json

api_root_url = "https://musicbrainz.org/ws/2/"
entity = "artist"
query = "paulo abucater"
url = f"{api_root_url}{entity}/?query={query}&fmt=json&limit=5"
response = requests.get(url)
data = json.loads(response.text)
print(data)
```

```python
facts_list = data.get("artists", [])
facts = "\n".join([f"{i+1}. {fact}\n" for i, fact in enumerate(facts_list)])
len(facts)
print(facts)
```
